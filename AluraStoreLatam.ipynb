### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

### ✅ Definir una función de exploración inicial por tienda

```python
def explorar_tienda(df, nombre_tienda = "Tienda"):
    print(f"\n--- {nombre_tienda}")
    
    print("Primeras filas:")
    display(df.head())
    
    print("\nInfo General:")
    print(df.info())
    
    print("\nValores nulos:")
    print(df.isnull().sum())
    
    print("\nEstadístiscas descriptivas")
    display(df.describe())
    
    print("\n Columnas únicas relevantes:")
    print("Categorías:", df['Categoría del Producto'].unique())
    print("Métodos de pago", df['Método de pago'].unique())
    print("Rango de fechas", df['Fecha de Compra'].min(), "->", df['Fecha de Compra'].max())    
    
```

### Ejecutar la funcion en todas las tiendas

# 1. Análisis de facturación



```python
# Definir una función para calcular la facturación total de cada tienda

def ingreso_total_por_tienda(df, nombre_tienda = "Tienda"):
    ingreso = df['Precio'].sum()
    return nombre_tienda, ingreso
```

```python
# Convertir a un dataframe los ingresos de cada tienda
datos_ingresos = [
    ingreso_total_por_tienda(tienda, "Tienda 1"),
    ingreso_total_por_tienda(tienda2, "Tienda 2"),
    ingreso_total_por_tienda(tienda3, "Tienda 3"),
    ingreso_total_por_tienda(tienda4, "Tienda 4")
]

df_facturacion = pd.DataFrame(datos_ingresos, columns=["Tienda", "Ingreso Total"])
# Convertir los ingresos a millones para que sea más legible visualmente
df_facturacion["Ingreso en Millones"] = df_facturacion["Ingreso Total"] / 1e6
# Calcular el total de ingresos
total_ingreso = df_facturacion['Ingreso Total'].sum()

# Calcular el porcentaje de cada tienda
df_facturacion['Porcentaje del Total'] = ((df_facturacion['Ingreso Total'] / total_ingreso) * 100).round(2)
display(df_facturacion)
```

# 2. Ventas por categoría

```python
# Definir una función para calcular las ventas por categoría

def ventas_por_categoria(df, nombre_tienda):
    ventas = df['Categoría del Producto'].value_counts().reset_index()
    ventas.columns = ['Categoría del Producto', 'Cantidad Vendida']
    ventas["Tienda"] = nombre_tienda
    return ventas
```

```python
# Crear lista de DataFrames con la tienda incluida
lista_df_ventas_categoria = [

    ventas_por_categoria(tienda, "Tienda 1"),
    ventas_por_categoria(tienda2, "Tienda 2"),
    ventas_por_categoria(tienda3, "Tienda 3"),
    ventas_por_categoria(tienda4, "Tienda 4")
]

df_ventas_categoria = pd.concat(lista_df_ventas_categoria, ignore_index= True)
display(df_ventas_categoria)
```

# 3. Calificación promedio de la tienda


```python
# ✅ Definir una función para calificar el promedio por tienda

def calificacion_promedio(df, nombre_tienda = "tienda"):
    promedio = df['Calificación'].mean()
    print(f"Calificación promedio en {nombre_tienda}: {promedio:.2f}")
    return nombre_tienda, promedio
```

```python
# Creando un Dataframe para calificación promedio:
calificaciones_promedio = [
    calificacion_promedio(tienda, "Tienda 1"),
    calificacion_promedio(tienda2, "Tienda 2"),
    calificacion_promedio(tienda3, "Tienda 3"),
    calificacion_promedio(tienda4, "Tienda 4")
]

df_calificacion_promedio = pd.DataFrame(calificaciones_promedio, columns=["Tienda", "Promedio"]).round(2)
display(df_calificacion_promedio)
```

# 4. Productos más y menos vendidos

```python
# ✅ Definir una función para calcular los productos más y menos vendidos

def productos_top_y_flop(df, nombre_tienda, top_n=5):
    conteo = df['Producto'].value_counts().reset_index()
    conteo.columns = ['Producto', 'Cantidad']
    
    top = conteo.head(top_n).copy()
    top['Tipo'] = 'Más vendidos'
    top['Tienda'] = nombre_tienda

    flop = conteo.tail(top_n).copy()
    flop['Tipo'] = 'Menos vendidos'
    flop['Tienda'] = nombre_tienda

    return pd.concat([top, flop])
```

```python
# ✅ Uniendo todos los dataframe de cada tienda en uno solo

lista_productos = [
    productos_top_y_flop(tienda, "Tienda 1"),
    productos_top_y_flop(tienda2, "Tienda 2"),
    productos_top_y_flop(tienda3, "Tienda 3"),
    productos_top_y_flop(tienda4, "Tienda 4")
]

df_conteo_productos = pd.concat(lista_productos, ignore_index=True)
display(df_conteo_productos)
```

```python
# Calcular el porcentaje que representa cada producto con respecto a las ventas
# Paso 1: Calcular total de productos vendidos por tienda
ventas_totales_por_tienda = df_conteo_productos.groupby("Tienda")["Cantidad"].sum().reset_index()
ventas_totales_por_tienda.columns = ["Tienda", "Total_Tienda"]

# Paso 2: Unir el total con el DataFrame de conteo de productos
df_conteo_productos = df_conteo_productos.merge(ventas_totales_por_tienda, on="Tienda")

# Paso 3: Calcular el porcentaje
df_conteo_productos["Porcentaje"] = (df_conteo_productos["Cantidad"] / df_conteo_productos["Total_Tienda"]) * 100

# Opcional: Redondear el porcentaje
df_conteo_productos["Porcentaje"] = df_conteo_productos["Porcentaje"].round(2)

# Mostrar resultado
display(df_conteo_productos)
```

# 5. Envío promedio por tienda

```python
# ✅ Definir una función para calcular el valor promedio por tienda

def costo_envio_promedio(df, nombre_tienda = "tienda"):
    costo_promedio = df['Costo de envío'].mean()
    print(f"Costo de envío promedio en {nombre_tienda}: $ {costo_promedio:.2f}")
    return nombre_tienda, costo_promedio
```

```python
# ✅ Creando un dataframe con el costo promedio por tienda

costo_promedio = [
    costo_envio_promedio(tienda, "Tienda 1"),
    costo_envio_promedio(tienda2, "Tienda 2"),
    costo_envio_promedio(tienda3, "Tienda 3"),
    costo_envio_promedio(tienda4, "Tienda 4")
]

df_costo_promedio = pd.DataFrame(costo_promedio, columns=["Tienda", "Costo Promedio"]).round(2)
display(df_costo_promedio)
```

### 🔝 4. Gráfico de barras horizontales: Top 5 productos más vendidos(por tienda)

```python
# Filtrar solo los más vendidos
top_5 = df_conteo_productos[df_conteo_productos['Tipo'] == 'Más vendidos']

# Obtener tiendas únicas
tiendas = top_5['Tienda'].unique()

# Crear figura con 2 filas y 2 columnas
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

# Iterar por cada tienda y su gráfico correspondiente
for i, tienda in enumerate(tiendas):
    data = top_5[top_5['Tienda'] == tienda]
    
    bars = axes[i].barh(data['Producto'], data['Cantidad'], color='Skyblue')
    axes[i].set_title(f'Top 5 productos - {tienda}')
    axes[i].invert_yaxis()  # Producto más vendido arriba
    axes[i].set_xlabel('Cantidad vendida')
    axes[i].set_ylabel('Producto')
    axes[i].grid(axis='x', linestyle='--', alpha=0.5)
    
    # Crear etiquetas con cantidad y porcentaje
    etiquetas = [f"{int(c)} ({p:.1f}%)" for c, p in zip(data['Cantidad'], data['Porcentaje'])]
    axes[i].bar_label(bars, labels=etiquetas, padding=5)

# Ajustar diseño
plt.tight_layout()
plt.show()
```
